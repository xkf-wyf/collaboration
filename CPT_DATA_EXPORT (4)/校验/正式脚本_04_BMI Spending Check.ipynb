```sql
-- BMI Spending raw_data, RTB Spending与看板ads层对数，By store+category+brand+product_line_group 
create or replace table ${spark.var.catalog.tmp}.fin.tmp_bmi_spending_rtb_check
select 
    coalesce(t1.month,t2.month) as month
    ,coalesce(t1.bu_name,t2.bu) as bu
    ,coalesce(t1.store_name,t2.store_name) as store_name
    ,coalesce(t1.category,t2.category) as category_name
    ,coalesce(t1.brand,t2.brand) as brand_name
    ,coalesce(t1.product_line_group,t2.product_line_group) as product_line_group
    ,coalesce(rtb_spending_1,0) as rtb_spending_cpt
    ,coalesce(rtb_spending_2,0) as rtb_spending_golden
    ,int(rtb_spending_cpt) - int(rtb_spending_golden) as gap_absolute
    ,coalesce(round(gap_absolute / int(rtb_spending_cpt),8), if(gap_absolute!=0,1,if(gap_absolute=0,0,null))) as `gap%`
    ,case when gap_absolute != 0 and (
            (coalesce(t1.category,t2.category) = 'SKIN' and coalesce(t1.brand,t2.brand) in ('DOVE','PONDS'))
            or (coalesce(t1.brand,t2.brand) in ('TIGI','HAZELINE'))
          ) then '黄金公式过滤品牌'
          when gap_absolute != 0 and int(sum(coalesce(rtb_spending_1,0) - coalesce(rtb_spending_2,0)) 
                                    over(partition by coalesce(t1.month,t2.month),coalesce(t1.bu_name,t2.bu),coalesce(t1.store_name,t2.store_name)
                                                      ,coalesce(t1.category,t2.category),coalesce(t1.brand,t2.brand))) = 0 
               and max(if(coalesce(t1.product_line_group,t2.product_line_group)='GWP/SAMPLE',1,0)) 
                    over(partition by coalesce(t1.month,t2.month),coalesce(t1.bu_name,t2.bu),coalesce(t1.store_name,t2.store_name)
                                  ,coalesce(t1.category,t2.category),coalesce(t1.brand,t2.brand)) = 1
              then '黄金公式小样处理成GWP/SAMPLE'
          when gap_absolute != 0 and t1.product_line_group is null 
               and coalesce(t2.product_line_group,'') != 'GWP/SAMPLE' then '空品线组黄金公式过滤'
          when gap_absolute != 0 then '原因待查'
          end as reason
from (


  select month,bu_name,store_name,category_name as category,brand_name as brand,product_line_group 
  ,round(sum(value_taxed),2) as rtb_spending_1 --逻辑，取value_taxed含税字段
  --,round(sum(value),2) as rtb_spending_1
  from ${spark.var.catalog.ads}.fin_cpt.ads_bmi_spending_mf
  where dt = '$dt' and index_type = 'Paid Media-RTB'
  group by month,bu_name,store_name,category_name,brand_name,product_line_group  


) t1
full join (


  select month,store_name,bu,category,brand,product_line_group,sum(value) as rtb_spending_2
  from (
  select
      replace(period,'-','')                        as month
      -- ,store_group                                  as store_name
      -- ,case when store_group='TMALL SUPER' then 'TMALL SUPER - UL' 
      --       --  when store_group='JD - REGENERATE' then 'JD - UL'
      -- else store_group end as store_name --20260421修改 
      ,store_group  as store_name --20260907
      ,bu as bu
      ,category                                     as category
      ,brand                                        as brand
      ,product_group                                as product_line_group
      ,t2.column_name                               as index_type
      ,sum(value)                                   as value
      ,'Automatic'                                  as create_by
      ,date_format(from_utc_timestamp(current_timestamp,'Asia/Shanghai'), 'yyyy-MM-dd HH:mm:ss') as create_time
  from ${spark.var.catalog.ads}.ddm_dset.ads_ec_g_fomular_ndy_traffic_fact t1 
  left join ${spark.var.catalog.ads}.ddm_dset.ads_ec_g_fomular_dim_traffic_metric_cfg t2
  on t1.metric_id = t2.metric_id
  where period_type = 'Month' and period between '$start_mth' and '$end_mth'
  group by period,store_group,bu,category,brand,product_group,t2.column_name
  having index_type is not null
  order by month,store_name,bu,category,brand,product_line_group,index_type
  )
  where index_type = 'rtb_spending' 
  group by month,store_name,bu,category,brand,product_line_group    


  
) t2
on t1.month = t2.month and t1.bu_name = t2.bu and t1.store_name = t2.store_name and t1.category = t2.category and t1.brand = t2.brand
and t1.product_line_group = t2.product_line_group
where 1=1
--and coalesce(t1.category,t2.category) in ('SKIN','HAIR')
and coalesce(t1.bu_name,t2.bu) in ('BEAUTY&WELLBEING','HOME CARE','PERSONAL CARE')
and coalesce(t1.store_name,t2.store_name) in ($store)
order by coalesce(t1.month,t2.month),coalesce(t1.bu_name,t2.bu),coalesce(t1.store_name,t2.store_name),coalesce(t1.category,t2.category),coalesce(t1.brand,t2.brand)
        ,coalesce(t1.product_line_group,t2.product_line_group)
;
```

```sql
-- BMI Spending raw_data, pfm_spending和底表By store+category+brand+plg 
create or replace table ${spark.var.catalog.tmp}.fin.tmp_bmi_spending_pfm_check
select 
     coalesce(t1.month,t2.month) as month
    ,coalesce(t1.index_type,t2.index_type) as pfm_type
    ,coalesce(t1.bu_name,t2.bu_name)  as bu
    ,coalesce(t1.store_name,t2.store_name) as store_name
    ,coalesce(t1.category,t2.category) as category_name
    ,coalesce(t1.brand,t2.brand) as brand_name
    ,coalesce(t1.product_line_group,t2.product_line_group) as product_line_group
    ,pfm_spending_1 as pfm_spending_cpt
    ,pfm_spending_2 as pfm_spending_source
    ,int(coalesce(pfm_spending_1,0) - coalesce(pfm_spending_2,0)) as gap_amount
    ,coalesce(round(gap_amount / int(pfm_spending_2),8), if(gap_amount!=0,1,if(gap_amount=0,0,null))) as `gap%`
from (


  select month,index_type,bu_name,store_name,category_name as category,brand_name as brand,product_line_group
  ,sum(value_taxed) as pfm_spending_1  --逻辑，取value_taxed含税字段
  --,sum(value) as pfm_spending_1
  from ${spark.var.catalog.ads}.fin_cpt.ads_bmi_spending_mf
  where dt = '$dt' and index_type != 'Paid Media-CPS'
  group by month,index_type,bu_name,store_name,category_name,brand_name,product_line_group



) t1
left join (



select
    month
    ,bu                                           as bu_name      
    ,case when customer ='JD - REGENERATE' then 'JD - UL'  --20260907
          else customer end  as store_name  
    ,category  
    ,brand
    ,product_line_group                            
    ,concat('Paid Media-',pfm_type)               as index_type           
    ,sum(pfm_spending)   as pfm_spending_2                         
from (
  -- 能匹配上打标表的用打标表的维度
  select  
      replace(substr(t1.period_key,1,7),'-','')   as month                                
      ,if(t1.period_key > '2025-10-16' and t2.shop_id = '621699362', 'PDD - VSL', t3.customer) as customer     -- 20251204修改          
      ,t2.category                as category              
      ,t2.brand                   as brand 
      ,t2.bu                      as bu         
      ,t2.product_line_group      as product_line_group
      ,t1.pfm_type                as pfm_type
      ,t1.pfm_spending            as pfm_spending
  from ${spark.var.catalog.dws}.ddm.dws_pfm_total_item_df t1
  left join (
    select *
    from ${spark.var.catalog.dim}.ddm.dim_prd_item
    where is_current = 1
  ) t2
  on t1.item_id = t2.item_id and if(t1.platform in ('VIP','TMALL SUPER'),t1.shop_id,t1.platform) = if(t2.platform in ('VIP','TMALL SUPER'),t2.shop_id,t2.platform)
  and t1.period_key >= t2.start_time and t1.period_key <= t2.end_time
  left join (
    select *
    from ${spark.var.catalog.dim}.ddm.dim_store
    where status = 1
  ) t3
  on COALESCE(t2.shop_id,t1.shop_id) = t3.shop_id
  where substr(t1.period_key,1,7) between '${start_mth}' and '${end_mth}'
  and t2.bu is not null
  union all 
  -- 店铺维度的数直接用事实表的维度
  select  
      replace(substr(t1.period_key,1,7),'-','')   as month                                
      ,if(t1.period_key > '2025-10-16' and t1.shop_id = '621699362', 'PDD - VSL', t3.customer) as customer     -- 20251204修改              
      ,t1.pfm_category                  as category          
      ,t1.pfm_brand                     as brand
      ,t1.pfm_bu                        as bu             
      ,t1.pfm_product_line_group        as product_line_group
      ,t1.pfm_type                      as pfm_type
      ,t1.pfm_spending                  as pfm_spending
  from ${spark.var.catalog.dws}.ddm.dws_pfm_total_item_df t1
  left join (
    select *
    from ${spark.var.catalog.dim}.ddm.dim_prd_item
    where is_current = 1
  ) t2
  on t1.item_id = t2.item_id and if(t1.platform in ('VIP','TMALL SUPER'),t1.shop_id,t1.platform) = if(t2.platform in ('VIP','TMALL SUPER'),t2.shop_id,t2.platform)
  and t1.period_key >= t2.start_time and t1.period_key <= t2.end_time
  left join (
    select *
    from ${spark.var.catalog.dim}.ddm.dim_store
    where status = 1
  ) t3
  on t1.shop_id = t3.shop_id
  where substr(t1.period_key,1,7) between '${start_mth}' and '${end_mth}'
  and t2.bu is null
  and (t1.item_id in ('0','-') or (t1.pfm_type = 'Package' and t1.item_id is null))  --店铺维度和京东的0和-,Package的null
  and coalesce(t1.pfm_product_line,'') != '-'       --剔除JD的品线为-的,归不到品线组
)
where  1=1
group by month,category,brand,product_line_group,pfm_type,bu
,case when customer ='JD - REGENERATE' then 'JD - UL'  else customer end  --20260907
having coalesce(pfm_type,'') != ''




) t2
on t1.index_type = t2.index_type and t1.store_name = t2.store_name and t1.month = t2.month
and t1.category = t2.category and t1.brand = t2.brand and t1.bu_name = t2.bu_name -- 20260909  <=>
and (t1.product_line_group = t2.product_line_group
     or (t1.product_line_group is null and t2.product_line_group is null))
where 1=1
-- and coalesce(t1.category,t2.category,'') in ('SKIN','HAIR','')
-- and coalesce(t1.category,t2.category) in ('SKIN','HAIR')
and coalesce(t1.bu_name,t2.bu_name) in ('BEAUTY&WELLBEING','HOME CARE','PERSONAL CARE') --新增过滤
and coalesce(t1.store_name,t2.store_name) in ($store)
order by coalesce(t1.month,t2.month),coalesce(t1.index_type,t2.index_type),coalesce(t1.store_name,t2.store_name),coalesce(t1.bu_name,t2.bu_name)
        ,coalesce(t1.category,t2.category),coalesce(t1.brand,t2.brand),coalesce(t1.product_line_group,t2.product_line_group)
        
;
```

```sql
-- BMI Spending raw_data, cps_spending和底表By site+bu+store+category+brand+plg
create or replace table ${spark.var.catalog.tmp}.fin.tmp_bmi_spending_cps_check
select
    coalesce(t1.month,t2.month) as month
    ,coalesce(t1.site,t2.site) as site
    ,coalesce(t1.bu,t2.bu) as bu
    ,coalesce(t1.store_name,t2.store_name) as store_name
    ,coalesce(t1.category,t2.category) as category
    ,coalesce(t1.brand,t2.brand) as brand
    ,coalesce(t1.plg,t2.plg) as product_line_group
    ,cps_spending_1 as cps_spending_cpt
    ,cps_spending_2 as cps_pending_source
    ,int(coalesce(cps_spending_1,0) - coalesce(cps_spending_2,0)) as gap_amount
    ,coalesce(round(gap_amount / nullif(cps_spending_2,0),8), if(gap_amount!=0,1,if(gap_amount=0,0,null))) as `gap%`
from (
  select month
    ,trim(split_part(store_name,'-',1)) as site
    ,bu_name as bu
    ,store_name as store_name
    ,category_name as category
    ,brand_name as brand
    ,product_line_group as plg
    ,sum(value_taxed) as cps_spending_1 --逻辑，取value_taxed含税字段
    -- ,sum(value) as cps_spending_1
    from ${spark.var.catalog.ads}.fin_cpt.ads_bmi_spending_mf
    where dt = '$dt' and index_type = 'Paid Media-CPS'
    group by month
      ,trim(split_part(store_name,'-',1))
      ,bu_name
      ,store_name
      ,category_name
      ,brand_name
      ,product_line_group
) t1
full join (
  select replace(substr(t1.period_key,1,7),'-','') as month
    ,t1.platform as site
    ,t2.bu as bu
    ,t3.customer as store_name
    ,t2.category as category
    ,t2.brand as brand
    ,t2.product_line_group as plg
    ,sum(t1.spending) as cps_spending_2
  from ${spark.var.catalog.tmp}.fin.tmp_cps_spending t1
  left join (
    select *
    from ${spark.var.catalog.dim}.ddm.dim_prd_item
    where is_current = 1
    -- and end_time='9999-12-31'
  ) t2
  on t1.item_id = t2.item_id
  -- and t1.period_key >= t2.start_time and t1.period_key <= t2.end_time
  and replace(last_day( date_format(to_date(t1.period_key, 'yyyy-MM-dd'), 'yyyy-MM-dd') ) , '-', '') >=  replace(t2.start_time, '-', '')
  and replace(last_day( date_format(to_date(t1.period_key, 'yyyy-MM-dd'), 'yyyy-MM-dd') ) , '-', '') <=  replace(t2.end_time, '-', '') 
  left join (
    select *
    from ${spark.var.catalog.dim}.ddm.dim_store
    where status = 1
  ) t3
  on t2.shop_id = t3.shop_id
  where substr(t1.period_key,1,7) between '${start_mth}' and '${end_mth}'
  and t3.customer is not null
  group by replace(substr(t1.period_key,1,7),'-','')
    ,t1.platform
    ,t2.bu
    ,t3.customer
    ,t2.category
    ,t2.brand
    ,t2.product_line_group
) t2
on t1.month = t2.month 
   and t1.site = t2.site
   and coalesce(t1.bu,'') = coalesce(t2.bu,'')
   and coalesce(t1.store_name,'') = coalesce(t2.store_name,'')
   and coalesce(t1.category,'') = coalesce(t2.category,'')
   and coalesce(t1.brand,'') = coalesce(t2.brand,'')
   and coalesce(t1.plg,'') = coalesce(t2.plg,'')
order by coalesce(t1.month,t2.month)
    ,coalesce(t1.site,t2.site)
    ,coalesce(t1.bu,'')
    ,coalesce(t1.store_name,'')
    ,coalesce(t1.category,'')
    ,coalesce(t1.brand,'')
    ,coalesce(t1.plg,'')
;
```
